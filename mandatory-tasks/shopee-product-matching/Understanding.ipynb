# Shopee Product Matching


1) So we are given Listing of products as Titles-Images-label_groups we need
to find similar listings using product matching.

2) We are Doing product matching using 3 methods

    A) text-based product matching. :- We compare titles and find similar products

    B) image-based matching.:- We use images of products and compare them and find similar products

    C) Both Text and Image matching:- We get both text and image info and compare them and find similar products
